```python
from typing import List

class Address(BaseModel):
    street: str
    city: str
    state: str
    country: str

class Fellow(BaseModel):
    name: str
    score: int
    track: str
    address: List[Address]
```

```python
data = {
    "name": "Abdulmalik",
    "score": 90,
    "track": "AI Engineering",
    "address": [
        {"street": "Idi-Aba", "city": "Abeokuta", "state": "Ogun State", "country": "Nigeria"},
        {"street": "Iwo Road", "city": "Ibadan", "state": "Oyo State", "country": "Nigeria"}
    ]
}

fellow = Fellow(**data)
print(fellow)
```

```python
print(fellow.address[0].city)
print(fellow.address[1].city)
```

```python
print(fellow.json)
```

```python
print(fellow.model_dump_json())
```

```python
print(fellow.dict())
```

```python
print(fellow.model_dump())
```

```python
from datetime import datetime
from decimal import Decimal
from pydantic import BaseModel, Field, ValidationError
from typing import List

class ProductReview(BaseModel):
    review_id: int = Field(..., gt=0, lt=10000)
    usename: str = Field(..., min_length=3, max_length=12, pattern=r"^[A-Za-z0-9_]+$")
    scores: List[int] = Field(..., min_length=3, max_length=5)
    price: Decimal = Field(..., gt=0, max_digits=6, decimal_places=2)
    rating: float = Field(..., ge=0, le=5, multiple_of=0.5)
    is_active: bool = Field(default=True)
    created_at: datetime = Field(default_factory=datetime.now)
```

```python
def demo_int_field():
    print("Integer Field Validation")
    try:
        ProductReview(review_id=2, usename="alfamado", scores=[4, 5, 4], price=Decimal("10.00"), rating=4.5)
    except ValidationError as e:
        print(e)
```

```python
demo_int_field()
```

```python
def demo_str_field():
    print("String Field Validation")
    try:
        ProductReview(review_id=1, usename="Zooo", scores=[4, 5, 5], price=Decimal("20.00"), rating=4.0)
    except ValidationError as e:
        print(e)
```

```python
demo_str_field()
```

```python
def demo_list_field():
    print("\n List Field Validation")
    try:
        ProductReview(review_id=2, usename="techguru", scores=[5, 2, 2], price=Decimal("30.00"), rating=3.5)
    except ValidationError as e:
        print(e)
```

```python
demo_list_field()
```

```python
def demo_decimal_field():
    print("\n Decimal Field Validation")
    try:
        ProductReview(review_id=2, usename="damico", scores=[4, 4, 5], price=Decimal("5.50"), rating=4.5)
    except ValidationError as e:
        print(e)
```

```python
demo_decimal_field()
```

```python
def demo_float_field():
    print("\n Float Field Validation")
    try:
        ProductReview(review_id=4, usename="devguy", scores=[5, 4, 4], price=Decimal("99.99"), rating=4.0)
    except ValidationError as e:
        print(e)
```

```python
def demo_bool_field():
    print("\n Boolean Field Default")
    product = ProductReview(
        review_id=5,
        usename="jackson",
        scores=[5, 4, 5],
        price=Decimal("75.89"),
        rating=5.0
    )
    print("is_active =", product.is_active)
```

```python
demo_bool_field()
```

```python
def demo_datetime_field():
    print("\n Datetime Default Factory")
    product = ProductReview(
        review_id=5,
        username="romtime",
        scores=[3, 4, 5],
        price=Decimal("20.50"),
        rating=4.5
    )
    print("created_at =", product.created_at)
```
